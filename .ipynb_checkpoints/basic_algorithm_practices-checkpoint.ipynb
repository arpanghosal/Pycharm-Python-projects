```python
import numpy as np
```

```python
def binary_search(arr, target, precision=8): # Each step halves the search space → O(log⁡ n).
    left, right = 0, len(arr) - 1
    target = round(target, precision)  # Round the target
    while left <= right:
        mid = (left + right) // 2  # Find the middle index
        arr[mid] = round(arr[mid], precision)
        if arr[mid] == target:
            return mid  # Found target
        elif arr[mid] < target:
            left = mid + 1  # Search the right half
        else:
            right = mid - 1  # Search the left half
    return -1  # Not found
```
