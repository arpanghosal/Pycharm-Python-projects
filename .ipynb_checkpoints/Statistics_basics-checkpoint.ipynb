# **Error Function (mathrm{erf})**

The error function is defined as

$$
\mathrm{erf}(x) = \frac{2}{\sqrt{\pi}} \int_{0}^{x} e^{-t^2}\,dt.
$$

It quantifies the area under the Gaussian curve $(e^{-t^2})$ from 0 to x, scaled so that $(\mathrm{erf}(\infty) = 1)$.

**Key points:**
- (mathrm{erf}(0) = 0).  
- (mathrm{erf}(x)) rises monotonically toward 1 as x to $\infty$.

In statistics, for a standard normal variable Z,

$$
P\bigl(-x < Z < x\bigr) = \mathrm{erf}\!\bigl(\tfrac{x}{\sqrt{2}}\bigr).
$$


```python
def erf_numeric(x, steps=10000):
    """
    Approximate erf(x) = (2/√π) ∫₀ˣ e^{-t²} dt using the trapezoidal rule.
    """
    dx = x / steps
    total = 0.5 * (math.exp(-0**2) + math.exp(-x**2))
    for i in range(1, steps):
        t = i * dx
        total += math.exp(-t*t)
    return (2 / math.sqrt(math.pi)) * total * dx
```

```python
for x in [0.0, 0.5, 1.0, 2.0]:
    approx = erf_numeric(x)
    exact  = math.erf(x)
    print(f"x = {x:<3} | Approximate erf: {approx:.6f} | math.erf: {exact:.6f}")
```

# Evaluation of Z-score (single-tailed)

```python
def phi(x):
    """
    Standard normal cumulative distribution function (CDF).
    """
    return 0.5 * (1 + math.erf(x / math.sqrt(2)))

def inv_phi(p, tol=1e-6):
    """
    Inverse of the standard normal CDF using bisection.
    Approximates z such that phi(z) = p.
    """
    low, high = -10.0, 10.0
    while high - low > tol:
        mid = (low + high) / 2
        if phi(mid) < p:
            low = mid
        else:
            high = mid
    return (low + high) / 2
```

```python
#  Define Hypotheses and Example Parameters
# H0 (null hypothesis): μ = μ0
# H1 (alternative hypothesis): μ > μ0  (one-sided)
mu0 = 100      # Claimed population mean
x_bar = 105    # Observed sample mean
sigma = 15     # Known population standard deviation
n = 25         # Sample size
```

```python
# Cell 3: Compute the Test Statistic (Z)
z_stat = (x_bar - mu0) / (sigma / math.sqrt(n))
print(f"Z-statistic: {z_stat:.4f}")
```

```python
# Cell 4: Compute One-Sided P-value
p_value = 1 - phi(z_stat)
print(f"One-sided p-value: {p_value:.6f}")
```

```python
# Cell 5: Compute Significance (σ-level) from P-value
# For a one-sided test, significance z = inv_phi(1 - p_value)
sig_z = inv_phi(1 - p_value)
print(f"Significance (sigma level): {sig_z:.4f}")
```
