```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import make_blobs
```

```python
# Generate 2D blobs
X, _ = make_blobs(n_samples=300, centers=1, n_features=2, cluster_std=[5])
```

```python
X
```

```python
# Fit PCA
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)
```

```python
pca
```

```python

# Reconstruct to 2D for visualization
X_proj = pca.inverse_transform(X_pca)
```

```python
X_proj
```

```python

# Plot original and projected data
plt.scatter(X[:, 0], X[:, 1], alpha=0.3, label="Original")
plt.scatter(X_proj[:, 0], X_proj[:, 1], alpha=0.8, label="PCA Projection", color='r')
plt.legend()
plt.title("PCA: 2D to 1D and back")
plt.axis('equal')
plt.grid()
plt.show()


#The red points lie along the principal axis (1D projection)
#PCA found the axis of maximum variance
```

# Real data analysis

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python

data = load_iris()
X = data.data
y = data.target
```

```python
# Standardize features before PCA
X_std = StandardScaler().fit_transform(X)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_std)

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y)
plt.xlabel('PC 1')
plt.ylabel('PC 2')
plt.title("PCA on Iris Dataset (2D projection)")
plt.colorbar(label='Class label')
plt.grid()
plt.show()

#We’ve reduced 4-dimensional data to 2D while still being able to separate most of the classes.
```
